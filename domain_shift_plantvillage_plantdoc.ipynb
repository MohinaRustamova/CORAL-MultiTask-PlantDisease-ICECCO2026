```python
import random
from PIL import Image
import matplotlib.pyplot as plt

IMG_EXTS = ('.jpg', '.jpeg', '.png', '.bmp', '.tif', '.tiff')

# --------------------------------------------------
# 4. Pick 5 labels that have images in BOTH datasets
#    (using the counts you already computed above)
# --------------------------------------------------
labels_with_both = [
    lbl for lbl in common_labels
    if pv_val_counts.get(lbl, 0) > 0 and pd_counts.get(lbl, 0) > 0
]

if len(labels_with_both) < 5:
    raise ValueError(f"Only {len(labels_with_both)} labels exist in both datasets with images.")

# score = how many images we have in the "smaller" dataset for that label
label_scores = {
    lbl: min(pv_val_counts.get(lbl, 0), pd_counts.get(lbl, 0))
    for lbl in labels_with_both
}

labels_sorted = sorted(labels_with_both, key=lambda x: label_scores[x], reverse=True)
selected_labels = labels_sorted[:5]

print("Selected labels for the figure:")
for lbl in selected_labels:
    print(f"- {lbl} (PV val: {pv_val_counts.get(lbl,0)}, PD test: {pd_counts.get(lbl,0)})")

# --------------------------------------------------
# 5. Build file lists for each selected label
# --------------------------------------------------
def get_pv_file_map(base_dir, labels):
    file_map = {lbl: [] for lbl in labels}
    for cls in os.listdir(base_dir):
        cls_path = os.path.join(base_dir, cls)
        if not os.path.isdir(cls_path) or cls not in labels:
            continue
        for f in os.listdir(cls_path):
            if f.lower().endswith(IMG_EXTS):
                file_map[cls].append(os.path.join(cls_path, f))
    return file_map

def get_pd_file_map(base_dir, pd_to_pv_map, labels):
    file_map = {lbl: [] for lbl in labels}
    for pd_cls, pv_cls in pd_to_pv_map.items():
        if pv_cls not in labels:
            continue
        cls_path = os.path.join(base_dir, pd_cls)
        if not os.path.isdir(cls_path):
            continue
        for f in os.listdir(cls_path):
            if f.lower().endswith(IMG_EXTS):
                file_map[pv_cls].append(os.path.join(cls_path, f))
    return file_map

pv_files = get_pv_file_map(pv_val_dir, selected_labels)    # from your previous cell
pd_files = get_pd_file_map(pd_test_dir, pd_to_pv_map, selected_labels)

# --------------------------------------------------
# 6. Sample 1 image per dataset per label (total 10)
# --------------------------------------------------
random.seed(42)  # for reproducibility

pairs = []  # list of (label, pvd_path, pd_path)
for lbl in selected_labels:
    if len(pv_files[lbl]) == 0 or len(pd_files[lbl]) == 0:
        raise ValueError(f"No images for label {lbl} in one of the datasets.")
    pvd_img_path = random.choice(pv_files[lbl])
    pd_img_path  = random.choice(pd_files[lbl])
    pairs.append((lbl, pvd_img_path, pd_img_path))

# --------------------------------------------------
# 7. Plot a 5×2 grid and save the figure
# --------------------------------------------------
n_labels = len(pairs)
fig, axes = plt.subplots(n_labels, 2, figsize=(6, 3 * n_labels))

if n_labels == 1:
    axes = [axes]

for i, (lbl, pvd_path, pd_path) in enumerate(pairs):
    ax_pv, ax_pd = axes[i]

    img_pv = Image.open(pvd_path).convert("RGB")
    img_pd = Image.open(pd_path).convert("RGB")

    ax_pv.imshow(img_pv)
    ax_pv.axis("off")
    ax_pv.set_title(f"PlantVillage – {lbl}", fontsize=10)

    ax_pd.imshow(img_pd)
    ax_pd.axis("off")
    ax_pd.set_title(f"PlantDoc – {lbl}", fontsize=10)

plt.tight_layout()
plt.savefig("pv_vs_pd_5labels_10images.png", dpi=300, bbox_inches="tight")
plt.show()

print("Saved figure as pv_vs_pd_5labels_10images.png")

# --------------------------------------------------
# 8. (Optional) copy the 10 exact images into a folder
#    so you can upload them into the paper
# --------------------------------------------------
import shutil

out_dir = "/kaggle/working/paper_examples_5labels"
os.makedirs(out_dir, exist_ok=True)

for i, (lbl, pvd_path, pd_path) in enumerate(pairs, start=1):
    pvd_ext = os.path.splitext(pvd_path)[1]
    pd_ext  = os.path.splitext(pd_path)[1]

    pvd_out = os.path.join(out_dir, f"{i:02d}_{lbl}_PlantVillage{pvd_ext}")
    pd_out  = os.path.join(out_dir, f"{i:02d}_{lbl}_PlantDoc{pd_ext}")

    shutil.copy(pvd_path, pvd_out)
    shutil.copy(pd_path,  pd_out)

print(f"Copied 10 images into: {out_dir}")
```

```python
import os
import torch
from torch import nn, optim
from torch.utils.data import DataLoader
from torchvision import datasets, transforms, models

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Device:", device)

# ------------------------------------------------
# 0. Helpers from previous cell
# ------------------------------------------------
def split_label(full_label):
    # "Apple___Apple_scab" -> ("Apple", "Apple_scab")
    return full_label.split("___")

# common_labels is already defined earlier in your notebook
# we now build DISEASE label space from these
disease_names = sorted({split_label(lbl)[1] for lbl in common_labels})
disease_to_idx = {d: i for i, d in enumerate(disease_names)}
idx_to_disease = {i: d for d, i in disease_to_idx.items()}

print("Disease classes:", disease_names)
print("#Diseases:", len(disease_names))


# -----------------------------
# 1. TRANSFORMS
# -----------------------------
train_tf = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(20),
    transforms.ColorJitter(brightness=0.2, contrast=0.2),
    transforms.ToTensor(),
])

val_tf = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
])

# -----------------------------
# 2. LOAD DATASETS (disease labels)
# -----------------------------
def get_disease_dataset(base_dir, full_labels, transform, disease_to_idx):
    """
    base_dir: root with PlantVillage-style folders (e.g. 'Apple___Apple_scab')
    full_labels: list of allowed full labels (PlantVillage style, e.g. common_labels)
    returns: CustomDataset, disease_to_idx
    """
    label_folders = [
        os.path.join(base_dir, cls)
        for cls in os.listdir(base_dir)
        if cls in full_labels and os.path.isdir(os.path.join(base_dir, cls))
    ]

    IMG_EXTS = ('.jpg', '.jpeg', '.png')
    files = []
    targets = []

    for cls_path in label_folders:
        full_label = os.path.basename(cls_path)          # e.g. "Apple___Apple_scab"
        _, disease_name = split_label(full_label)        # -> "Apple_scab"
        disease_id = disease_to_idx[disease_name]        # map to int

        for fname in os.listdir(cls_path):
            if fname.lower().endswith(IMG_EXTS):
                files.append(os.path.join(cls_path, fname))
                targets.append(disease_id)

    # Custom dataset
    from PIL import Image

    class DiseaseDataset(torch.utils.data.Dataset):
        def __init__(self, files, targets, transform):
            self.files = files
            self.targets = targets
            self.transform = transform

        def __len__(self):
            return len(self.files)

        def __getitem__(self, idx):
            img = Image.open(self.files[idx]).convert("RGB")
            if self.transform is not None:
                img = self.transform(img)
            label = self.targets[idx]       # disease_id
            return img, label

    return DiseaseDataset(files, targets, transform)

train_dir = "/kaggle/input/plantvillage/PlantVillage/train"
val_dir   = "/kaggle/input/plantvillage/PlantVillage/val"

train_ds = get_disease_dataset(train_dir, common_labels, train_tf, disease_to_idx)
val_ds   = get_disease_dataset(val_dir,   common_labels, val_tf,   disease_to_idx)

num_classes = len(disease_to_idx)
print("Number of DISEASE classes:", num_classes)

train_loader = DataLoader(train_ds, batch_size=128, shuffle=True, num_workers=2)
val_loader   = DataLoader(val_ds,   batch_size=32, shuffle=False, num_workers=2)

# -----------------------------
# 3. MODEL: MobileNetV3-Large
# -----------------------------
model = models.mobilenet_v3_large(weights="MobileNet_V3_Large_Weights.DEFAULT")
model.classifier[3] = nn.Linear(model.classifier[3].in_features, num_classes)
model = model.to(device)

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=1e-4)

# -----------------------------
# 4. ACCURACY FUNCTION
# -----------------------------
def accuracy(loader):
    model.eval()
    correct = total = 0
    with torch.no_grad():
        for x, y in loader:
            x, y = x.to(device), y.to(device)
            preds = model(x).argmax(1)
            correct += (preds == y).sum().item()
            total += y.size(0)
    return correct / total

# -----------------------------
# 5. TRAINING LOOP
# -----------------------------
EPOCHS = 20  # increase for real training

for epoch in range(EPOCHS):
    model.train()
    running_loss = 0.0
    for images, labels in train_loader:
        images, labels = images.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = model(images)           # [B, num_diseases]
        loss = criterion(outputs, labels) # labels are disease_ids
        loss.backward()
        optimizer.step()

        running_loss += loss.item()

    train_acc = accuracy(train_loader)
    val_acc   = accuracy(val_loader)
    print(f"Epoch {epoch+1}/{EPOCHS} | "
          f"Loss: {running_loss/len(train_loader):.4f} | "
          f"Train Acc: {train_acc:.3f} | Val Acc: {val_acc:.3f}")

# -----------------------------
# 6. SAVE MODEL
# -----------------------------
torch.save(model.state_dict(), "mobilenetv3_plantvillage_disease_only.pth")
print("Model saved as mobilenetv3_plantvillage_disease_only.pth")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix
import torch

def compute_confusion_matrix(model, loader, device, num_classes):
    model.eval()
    all_preds = []
    all_labels = []

    with torch.no_grad():
        for images, labels in loader:
            images, labels = images.to(device), labels.to(device)
            logits = model(images)
            preds = logits.argmax(1)

            all_preds.append(preds.cpu().numpy())
            all_labels.append(labels.cpu().numpy())

    all_preds = np.concatenate(all_preds)
    all_labels = np.concatenate(all_labels)

    cm = confusion_matrix(all_labels, all_preds, labels=list(range(num_classes)))
    return cm

def plot_confusion_matrix(cm, class_names, title, filename, figsize=(10,10)):
    plt.figure(figsize=figsize)
    plt.imshow(cm, interpolation='nearest', cmap='Blues')
    plt.title(title, fontsize=14)
    plt.colorbar()

    tick_marks = np.arange(len(class_names))
    plt.xticks(tick_marks, class_names, rotation=90, fontsize=6)
    plt.yticks(tick_marks, class_names, fontsize=6)

    plt.tight_layout()
    plt.xlabel("Predicted", fontsize=12)
    plt.ylabel("True", fontsize=12)
    plt.savefig(filename, dpi=300, bbox_inches='tight')
    plt.show()
    print(f"Saved: {filename}")
```

```python
import numpy as np
from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
import torch

# -----------------------------
# Helper function
# -----------------------------
def compute_binary_cm(model, loader, device, healthy_idx):
    """
    Computes 2-class confusion matrix:
      0 = healthy
      1 = diseased (all non-healthy merged)
    """
    model.eval()
    all_bin_preds = []
    all_bin_labels = []

    with torch.no_grad():
        for images, labels in loader:
            images, labels = images.to(device), labels.to(device)

            logits = model(images)
            preds = logits.argmax(1)

            # Convert to binary classes
            bin_labels = (labels != healthy_idx).long()  # diseased = 1
            bin_preds  = (preds  != healthy_idx).long()  # diseased = 1

            all_bin_labels.append(bin_labels.cpu().numpy())
            all_bin_preds.append(bin_preds.cpu().numpy())

    all_bin_labels = np.concatenate(all_bin_labels)
    all_bin_preds  = np.concatenate(all_bin_preds)

    cm = confusion_matrix(all_bin_labels, all_bin_preds, labels=[0, 1])
    return cm

# -----------------------------
# Plot function (binary)
# -----------------------------
def plot_binary_cm(cm, title, filename):
    plt.figure(figsize=(5,5))
    plt.imshow(cm, interpolation='nearest', cmap='Blues')
    plt.title(title, fontsize=14)
    plt.colorbar()

    classes = ["Healthy", "Disease"]

    tick_marks = np.arange(2)
    plt.xticks(tick_marks, classes, fontsize=10)
    plt.yticks(tick_marks, classes, fontsize=10)

    plt.xlabel("Predicted", fontsize=12)
    plt.ylabel("True", fontsize=12)

    # Annotate cells
    thresh = cm.max() / 2.
    for i in range(2):
        for j in range(2):
            plt.text(
                j, i, format(cm[i, j], 'd'),
                ha="center", va="center",
                color="white" if cm[i, j] > thresh else "black",
                fontsize=12
            )

    plt.tight_layout()
    plt.savefig(filename, dpi=300, bbox_inches='tight')
    plt.show()
    print("Saved:", filename)
```

```python
import os
import torch
from torch import nn, optim
from torch.utils.data import DataLoader, Dataset
from torchvision import transforms, models
from PIL import Image

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Device:", device)

# ----------------------------------------------------
# 1. Build crop & disease label spaces from common_labels
# ----------------------------------------------------
def split_label(full_label):
    # "Apple___Apple_scab" -> ("Apple", "Apple_scab")
    return full_label.split("___")

crops = sorted({split_label(lbl)[0] for lbl in common_labels})
diseases = sorted({split_label(lbl)[1] for lbl in common_labels})

crop_to_idx = {c: i for i, c in enumerate(crops)}
idx_to_crop = {i: c for c, i in crop_to_idx.items()}

disease_to_idx = {d: i for i, d in enumerate(diseases)}
idx_to_disease = {i: d for d, i in disease_to_idx.items()}

num_crops = len(crops)
num_diseases = len(diseases)

print("Crops:", crops)
print("Diseases:", diseases)
print(f"#Crops: {num_crops}, #Diseases: {num_diseases}")

# ----------------------------------------------------
# 2. Transforms
# ----------------------------------------------------
train_tf = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(20),
    transforms.ColorJitter(brightness=0.2, contrast=0.2),
    transforms.ToTensor(),
])

val_tf = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
])

# ----------------------------------------------------
# 3. Multi-task PlantVillage dataset (crop + disease)
# ----------------------------------------------------
class MultiTaskPlantVillageDataset(Dataset):
    def __init__(self, base_dir, common_labels, crop_to_idx, disease_to_idx, transform=None):
        self.files = []
        self.crop_targets = []
        self.disease_targets = []
        self.transform = transform

        IMG_EXTS = ('.jpg', '.jpeg', '.png')

        # Each folder is PlantVillage-style: "Crop___Disease"
        for cls in os.listdir(base_dir):
            cls_path = os.path.join(base_dir, cls)
            if not os.path.isdir(cls_path):
                continue
            if cls not in common_labels:
                continue

            crop_name, disease_name = split_label(cls)
            crop_id = crop_to_idx[crop_name]
            disease_id = disease_to_idx[disease_name]

            for fname in os.listdir(cls_path):
                if fname.lower().endswith(IMG_EXTS):
                    fpath = os.path.join(cls_path, fname)
                    self.files.append(fpath)
                    self.crop_targets.append(crop_id)
                    self.disease_targets.append(disease_id)

        print(f"Loaded {len(self.files)} images from {base_dir}")

    def __len__(self):
        return len(self.files)

    def __getitem__(self, idx):
        img_path = self.files[idx]
        img = Image.open(img_path).convert("RGB")
        if self.transform is not None:
            img = self.transform(img)
        crop_label = self.crop_targets[idx]
        disease_label = self.disease_targets[idx]
        return img, crop_label, disease_label

train_dir = "/kaggle/input/plantvillage/PlantVillage/train"
val_dir   = "/kaggle/input/plantvillage/PlantVillage/val"

train_ds = MultiTaskPlantVillageDataset(train_dir, common_labels, crop_to_idx, disease_to_idx, transform=train_tf)
val_ds   = MultiTaskPlantVillageDataset(val_dir,   common_labels, crop_to_idx, disease_to_idx, transform=val_tf)

train_loader = DataLoader(train_ds, batch_size=128, shuffle=True, num_workers=2)
val_loader   = DataLoader(val_ds,   batch_size=32, shuffle=False, num_workers=2)

from torchvision import models
import torch.nn as nn
import torch.optim as optim

# ----------------------------------------------------
# Corrected Multi-task MobileNetV3-Large
# ----------------------------------------------------
class MobileNetV3_MTL(nn.Module):
    def __init__(self, num_crops, num_diseases):
        super().__init__()
        backbone = models.mobilenet_v3_large(weights="MobileNet_V3_Large_Weights.DEFAULT")

        # IMPORTANT: feature dim is the INPUT to classifier[0], not classifier[3]
        feat_dim = backbone.classifier[0].in_features   # 960

        # remove classifier, keep feature extractor
        backbone.classifier = nn.Identity()
        self.backbone = backbone

        # two heads on top of shared features
        self.crop_head = nn.Linear(feat_dim, num_crops)
        self.disease_head = nn.Linear(feat_dim, num_diseases)

    def forward(self, x):
        feats = self.backbone(x)               # [B, 960]
        crop_logits = self.crop_head(feats)    # [B, num_crops]
        disease_logits = self.disease_head(feats)  # [B, num_diseases]
        return crop_logits, disease_logits

# recreate model & optimizer
model_mtl = MobileNetV3_MTL(num_crops=num_crops, num_diseases=num_diseases).to(device)

criterion_crop = nn.CrossEntropyLoss()
criterion_disease = nn.CrossEntropyLoss()
crop_loss_weight = 0.01
disease_loss_weight = 1.0

optimizer = optim.Adam(model_mtl.parameters(), lr=1e-4)


# ----------------------------------------------------
# 6. Accuracy helpers
# ----------------------------------------------------
def mtl_accuracy(loader):
    model_mtl.eval()
    correct_crop = total_crop = 0
    correct_disease = total_disease = 0

    with torch.no_grad():
        for images, crop_labels, disease_labels in loader:
            images = images.to(device)
            crop_labels = crop_labels.to(device)
            disease_labels = disease_labels.to(device)

            crop_logits, disease_logits = model_mtl(images)
            crop_preds = crop_logits.argmax(1)
            disease_preds = disease_logits.argmax(1)

            correct_crop += (crop_preds == crop_labels).sum().item()
            total_crop   += crop_labels.size(0)

            correct_disease += (disease_preds == disease_labels).sum().item()
            total_disease   += disease_labels.size(0)

    return (correct_crop / total_crop,
            correct_disease / total_disease)

# ----------------------------------------------------
# 7. Training loop (MTL)
# ----------------------------------------------------
EPOCHS = 20  # increase for serious training

for epoch in range(EPOCHS):
    model_mtl.train()
    running_loss = 0.0

    for images, crop_labels, disease_labels in train_loader:
        images = images.to(device)
        crop_labels = crop_labels.to(device)
        disease_labels = disease_labels.to(device)

        optimizer.zero_grad()

        crop_logits, disease_logits = model_mtl(images)

        loss_crop = criterion_crop(crop_logits, crop_labels)
        loss_disease = criterion_disease(disease_logits, disease_labels)

        loss = crop_loss_weight * loss_crop + disease_loss_weight * loss_disease
        loss.backward()
        optimizer.step()

        running_loss += loss.item()

    train_crop_acc, train_dis_acc = mtl_accuracy(train_loader)
    val_crop_acc, val_dis_acc     = mtl_accuracy(val_loader)

    print(f"Epoch {epoch+1}/{EPOCHS} | "
          f"Loss: {running_loss/len(train_loader):.4f} | "
          f"Train Crop Acc: {train_crop_acc:.3f} | Train Dis Acc: {train_dis_acc:.3f} | "
          f"Val Crop Acc: {val_crop_acc:.3f} | Val Dis Acc: {val_dis_acc:.3f}")

# ----------------------------------------------------
# 8. Save model
# ----------------------------------------------------
torch.save({
    "state_dict": model_mtl.state_dict(),
    "crop_to_idx": crop_to_idx,
    "disease_to_idx": disease_to_idx,
}, "mobilenetv3_mtl_crop_disease_pvd.pth")

print("MTL model saved as mobilenetv3_mtl_crop_disease_pvd.pth")
```

```python
import os
import torch
from torch import nn, optim
from torch.utils.data import DataLoader, Dataset
from torchvision import transforms, models
from PIL import Image
from itertools import cycle

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Device:", device)

# ----------------------------------------------------
# 0. Build crop & disease label spaces from common_labels
#     (assumes common_labels is defined earlier)
# ----------------------------------------------------
def split_label(full_label):
    return full_label.split("___")

crops = sorted({split_label(lbl)[0] for lbl in common_labels})
diseases = sorted({split_label(lbl)[1] for lbl in common_labels})

crop_to_idx = {c: i for i, c in enumerate(crops)}
disease_to_idx = {d: i for i, d in enumerate(diseases)}

num_crops = len(crops)
num_diseases = len(diseases)

print(f"#Crops: {num_crops}, #Diseases: {num_diseases}")

# Common image extensions
IMG_EXTS = ('.jpg', '.jpeg', '.png', '.bmp', '.tif', '.tiff')

# ----------------------------------------------------
# 1. Transforms
# ----------------------------------------------------
train_tf = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(20),
    transforms.ColorJitter(brightness=0.2, contrast=0.2),
    transforms.ToTensor(),
])

val_tf = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
])

# ----------------------------------------------------
# 2. Multi-task PlantVillage dataset
# ----------------------------------------------------
class MultiTaskPlantVillageDataset(Dataset):
    def __init__(self, base_dir, common_labels, crop_to_idx, disease_to_idx, transform=None):
        self.files, self.crop_targets, self.disease_targets = [], [], []
        self.transform = transform

        for cls in os.listdir(base_dir):
            cls_path = os.path.join(base_dir, cls)
            if not os.path.isdir(cls_path) or cls not in common_labels:
                continue
            crop_name, disease_name = split_label(cls)
            crop_id = crop_to_idx[crop_name]
            disease_id = disease_to_idx[disease_name]
            for fname in os.listdir(cls_path):
                if fname.lower().endswith(IMG_EXTS):
                    self.files.append(os.path.join(cls_path, fname))
                    self.crop_targets.append(crop_id)
                    self.disease_targets.append(disease_id)

        print(f"Loaded {len(self.files)} images from {base_dir}")

    def __len__(self):
        return len(self.files)

    def __getitem__(self, idx):
        img = Image.open(self.files[idx]).convert("RGB")
        if self.transform:
            img = self.transform(img)
        return img, self.crop_targets[idx], self.disease_targets[idx]

train_dir = "/kaggle/input/plantvillage/PlantVillage/train"
val_dir   = "/kaggle/input/plantvillage/PlantVillage/val"

train_ds = MultiTaskPlantVillageDataset(train_dir, common_labels, crop_to_idx, disease_to_idx, transform=train_tf)
val_ds   = MultiTaskPlantVillageDataset(val_dir,   common_labels, crop_to_idx, disease_to_idx, transform=val_tf)

train_loader = DataLoader(train_ds, batch_size=64, shuffle=True, num_workers=1)
val_loader   = DataLoader(val_ds, batch_size=32, shuffle=False, num_workers=1)

# ----------------------------------------------------
# 3. MobileNetV3 MTL model
# ----------------------------------------------------
class MobileNetV3_MTL(nn.Module):
    def __init__(self, num_crops, num_diseases):
        super().__init__()
        backbone = models.mobilenet_v3_large(weights="MobileNet_V3_Large_Weights.DEFAULT")
        feat_dim = backbone.classifier[0].in_features
        backbone.classifier = nn.Identity()
        self.backbone = backbone
        self.crop_head = nn.Linear(feat_dim, num_crops)
        self.disease_head = nn.Linear(feat_dim, num_diseases)

    def forward(self, x):
        feats = self.backbone(x)
        crop_logits = self.crop_head(feats)
        disease_logits = self.disease_head(feats)
        return crop_logits, disease_logits, feats  # return feats for CORAL

# ----------------------------------------------------
# 4. CORAL loss
# ----------------------------------------------------
def coral_loss(source_feats, target_feats):
    d = source_feats.size(1)
    src = source_feats - source_feats.mean(dim=0, keepdim=True)
    tgt = target_feats - target_feats.mean(dim=0, keepdim=True)
    cs = (src.t() @ src) / (source_feats.size(0) - 1)
    ct = (tgt.t() @ tgt) / (target_feats.size(0) - 1)
    return torch.mean((cs - ct) ** 2) / (4.0 * d * d)

# ----------------------------------------------------
# 5. Model, loss, optimizer
# ----------------------------------------------------
model_mtl = MobileNetV3_MTL(num_crops, num_diseases).to(device)
criterion_crop = nn.CrossEntropyLoss()
criterion_disease = nn.CrossEntropyLoss()
optimizer = optim.Adam(model_mtl.parameters(), lr=1e-4)

CORAL_LAMBDA = 0.1  # weight for CORAL loss
SRC_BATCH_SIZE = 4   # (not used directly but kept for reference)
TGT_BATCH_SIZE = 4

# ----------------------------------------------------
# 5a. Unlabeled PlantDoc dataset for CORAL
#      We treat PlantDoc test set as unlabeled target
# ----------------------------------------------------
pd_unlabeled_dir = "/kaggle/working/PlantDoc-Dataset/test"

use_coral = False
tgt_iter = None

if os.path.isdir(pd_unlabeled_dir):
    class UnlabeledPlantDocDataset(Dataset):
        def __init__(self, root, transform):
            self.files = []
            self.transform = transform

            # Walk all subfolders and collect images
            for cls in os.listdir(root):
                cls_path = os.path.join(root, cls)
                if not os.path.isdir(cls_path):
                    continue
                for fname in os.listdir(cls_path):
                    if fname.lower().endswith(IMG_EXTS):
                        self.files.append(os.path.join(cls_path, fname))

            print(f"Loaded {len(self.files)} unlabeled images from PlantDoc test for CORAL.")

        def __len__(self):
            return len(self.files)

        def __getitem__(self, idx):
            img = Image.open(self.files[idx]).convert("RGB")
            if self.transform:
                img = self.transform(img)
            return img

    tgt_ds = UnlabeledPlantDocDataset(pd_unlabeled_dir, val_tf)
    if len(tgt_ds) > 0:
        tgt_loader = DataLoader(tgt_ds, batch_size=TGT_BATCH_SIZE, shuffle=True, num_workers=1)
        tgt_iter = cycle(tgt_loader)
        use_coral = True
        print("CORAL enabled.")
    else:
        print("PlantDoc test folder has no images – CORAL disabled.")
else:
    print("pd_unlabeled_dir does not exist – CORAL disabled.")

# ----------------------------------------------------
# 6. Accuracy helper
# ----------------------------------------------------
def mtl_accuracy(loader):
    model_mtl.eval()
    correct_crop = total_crop = 0
    correct_dis = total_dis = 0
    with torch.no_grad():
        for images, crop_labels, disease_labels in loader:
            images = images.to(device)
            crop_labels = crop_labels.to(device)
            disease_labels = disease_labels.to(device)

            crop_logits, disease_logits, _ = model_mtl(images)
            correct_crop += (crop_logits.argmax(1) == crop_labels).sum().item()
            total_crop += crop_labels.size(0)

            correct_dis += (disease_logits.argmax(1) == disease_labels).sum().item()
            total_dis += disease_labels.size(0)

    return correct_crop / total_crop, correct_dis / total_dis

# ----------------------------------------------------
# 7. Training loop with CORAL
# ----------------------------------------------------
EPOCHS = 20

for epoch in range(EPOCHS):
    model_mtl.train()
    running_loss = 0.0

    for images, crop_labels, disease_labels in train_loader:
        images = images.to(device)
        crop_labels = crop_labels.to(device)
        disease_labels = disease_labels.to(device)

        optimizer.zero_grad()
        crop_logits, disease_logits, feats_src = model_mtl(images)

        loss_crop = criterion_crop(crop_logits, crop_labels)
        loss_disease = criterion_disease(disease_logits, disease_labels)
        loss = loss_crop + loss_disease

        # CORAL alignment with unlabeled PlantDoc
        if use_coral and tgt_iter is not None:
            tgt_imgs = next(tgt_iter).to(device)
            with torch.no_grad():
                _, _, feats_tgt = model_mtl(tgt_imgs)
            loss_coral = coral_loss(feats_src, feats_tgt)
            loss += CORAL_LAMBDA * loss_coral

        loss.backward()
        optimizer.step()
        running_loss += loss.item()

    train_crop_acc, train_dis_acc = mtl_accuracy(train_loader)
    val_crop_acc, val_dis_acc     = mtl_accuracy(val_loader)

    print(f"Epoch {epoch+1}/{EPOCHS} | "
          f"Loss: {running_loss/len(train_loader):.4f} | "
          f"Train Crop Acc: {train_crop_acc:.3f} | Train Dis Acc: {train_dis_acc:.3f} | "
          f"Val Crop Acc: {val_crop_acc:.3f} | Val Dis Acc: {val_dis_acc:.3f}")

# ----------------------------------------------------
# 8. Save model
# ----------------------------------------------------
torch.save({
    "state_dict": model_mtl.state_dict(),
    "crop_to_idx": crop_to_idx,
    "disease_to_idx": disease_to_idx,
}, "mobilenetv3_mtl_coral.pth")

print("MTL + CORAL model saved as mobilenetv3_mtl_coral.pth")
```

```python
import os
from PIL import Image
import torch
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
import torch.nn as nn

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Device:", device)

# -----------------------------
# 0. Helpers & disease label space
# -----------------------------
def split_label(full_label):
    return full_label.split("___")

disease_names = sorted({split_label(lbl)[1] for lbl in common_labels})
disease_to_idx = {d: i for i, d in enumerate(disease_names)}
idx_to_disease = {i: d for d, i in disease_to_idx.items()}
print("Disease classes:", disease_names)
print("#Diseases:", len(disease_names))

# -----------------------------
# 1. Transforms
# -----------------------------
val_tf = transforms.Compose([
    transforms.Resize((224,224)),
    transforms.ToTensor(),
])

# -----------------------------
# 2. PlantDoc dataset
# -----------------------------
class PlantDocDiseaseDataset(Dataset):
    def __init__(self, pd_dir, transform, pd_to_pv_map, disease_to_idx):
        self.files = []
        self.targets = []
        self.transform = transform
        IMG_EXTS = ('.jpg', '.jpeg', '.png')

        for folder in os.listdir(pd_dir):
            if folder not in pd_to_pv_map:
                continue
            pv_label = pd_to_pv_map[folder]
            if pv_label not in common_labels:
                continue
            _, disease_name = split_label(pv_label)
            if disease_name not in disease_to_idx:
                continue
            disease_id = disease_to_idx[disease_name]
            folder_path = os.path.join(pd_dir, folder)
            if not os.path.isdir(folder_path):
                continue
            for fname in os.listdir(folder_path):
                if fname.lower().endswith(IMG_EXTS):
                    self.files.append(os.path.join(folder_path, fname))
                    self.targets.append(disease_id)

        print(f"PlantDocDiseaseDataset: {len(self.files)} images")

    def __len__(self):
        return len(self.files)

    def __getitem__(self, idx):
        img = Image.open(self.files[idx]).convert("RGB")
        if self.transform:
            img = self.transform(img)
        label = self.targets[idx]
        return img, label

pd_test_dir = "/kaggle/working/PlantDoc-Dataset/test"
pd_dataset = PlantDocDiseaseDataset(pd_test_dir, val_tf, pd_to_pv_map, disease_to_idx)
pd_loader = DataLoader(pd_dataset, batch_size=16, shuffle=False, num_workers=1)

# -----------------------------
# 3. MTL model definition (same as Coral training)
# -----------------------------
class MobileNetV3_MTL(nn.Module):
    def __init__(self, num_crops, num_diseases):
        super().__init__()
        backbone = models.mobilenet_v3_large(
            weights="MobileNet_V3_Large_Weights.DEFAULT"
        )
        feat_dim = backbone.classifier[0].in_features
        backbone.classifier = nn.Identity()
        self.backbone = backbone
        self.crop_head = nn.Linear(feat_dim, num_crops)
        self.disease_head = nn.Linear(feat_dim, num_diseases)

    def forward(self, x):
        feats = self.backbone(x)
        crop_logits = self.crop_head(feats)
        disease_logits = self.disease_head(feats)
        return crop_logits, disease_logits, feats  # feats optional for CORAL

num_crops = 13  # same as training
num_diseases = len(disease_to_idx)
model_mtl = MobileNetV3_MTL(num_crops, num_diseases).to(device)

# -----------------------------
# 4. Load MTL + CORAL trained weights
# -----------------------------
ckpt = torch.load("mobilenetv3_mtl_coral.pth", map_location=device)
state_dict = ckpt.get("state_dict", ckpt)
model_mtl.load_state_dict(state_dict)
model_mtl.eval()
print("Loaded MTL + CORAL model successfully.")



# -----------------------------
# 5. EVALUATE DISEASE ACCURACY AND COLLECT PREDICTIONS
# -----------------------------
correct = total = 0
all_true_ids = []
all_pred_ids = []

with torch.no_grad():
    for images, labels in pd_loader:
        images, labels = images.to(device), labels.to(device)

        _, disease_logits, _ = model_mtl(images)
        preds = disease_logits.argmax(dim=1)

        correct += (preds == labels).sum().item()
        total += labels.size(0)

        all_true_ids.extend(labels.cpu().tolist())
        all_pred_ids.extend(preds.cpu().tolist())

acc = correct / total if total > 0 else 0.0
print(f"PlantDoc Test Disease Accuracy (MTL + CORAL): {acc:.3f} ({correct}/{total})")

# -----------------------------
# 6. SAVE CSV WITH PREDICTIONS
# -----------------------------
data = {
    'image_path': pd_dataset.files,
    'true_disease_id': all_true_ids,
    'pred_disease_id': all_pred_ids,
    'true_disease_name': [idx_to_disease[i] for i in all_true_ids],
    'pred_disease_name': [idx_to_disease[i] for i in all_pred_ids]
}
df = pd.DataFrame(data)
csv_path = 'plantdoc_disease_predictions_mtl_coral.csv'
df.to_csv(csv_path, index=False)
print(f"Saved CSV to {csv_path}")
```

# MTL with Coral for 3 task hierarchical

```python
import os
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, Dataset
from torchvision import transforms, models
from PIL import Image
from itertools import cycle

# 1. Setup Device
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Device:", device)

# ----------------------------------------------------
# 2. Labels and Mappings
# ----------------------------------------------------
common_labels = sorted([
    "Grape___healthy", "Corn_(maize)___Northern_Leaf_Blight", "Blueberry___healthy",
    "Tomato___Septoria_leaf_spot", "Corn_(maize)___Common_rust_", "Cherry_(including_sour)___healthy",
    "Apple___Cedar_apple_rust", "Tomato___Tomato_mosaic_virus", "Tomato___Late_blight",
    "Apple___Apple_scab", "Tomato___Early_blight", "Pepper,_bell___Bacterial_spot",
    "Peach___healthy", "Tomato___Bacterial_spot", "Apple___healthy", "Grape___Black_rot",
    "Potato___Early_blight", "Strawberry___healthy", "Tomato___healthy", "Pepper,_bell___healthy",
    "Potato___Late_blight", "Soybean___healthy", "Raspberry___healthy", "Squash___Powdery_mildew",
    "Tomato___Tomato_Yellow_Leaf_Curl_Virus", "Tomato___Leaf_Mold",
    "Corn_(maize)___Cercospora_leaf_spot Gray_leaf_spot"
])

def split_label(full_label):
    return full_label.split("___")

# Build label spaces
crops = sorted({split_label(lbl)[0] for lbl in common_labels})
diseases = sorted({split_label(lbl)[1] for lbl in common_labels})

crop_to_idx = {c: i for i, c in enumerate(crops)}
disease_to_idx = {d: i for i, d in enumerate(diseases)}
# Health index: 1 for Healthy, 0 for Diseased
# We identify "healthy" by checking if the disease string contains 'healthy'
health_to_idx = {"unhealthy": 0, "healthy": 1}

num_crops = len(crops)
num_diseases = len(diseases)
print(f"MTL Setup: {num_crops} Crops, 2 Health States, {num_diseases} Specific Diseases")

# Common image extensions
IMG_EXTS = ('.jpg', '.jpeg', '.png', '.bmp', '.tif', '.tiff')

# ----------------------------------------------------
# 3. Transforms
# ----------------------------------------------------
train_tf = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(20),
    transforms.ColorJitter(brightness=0.2, contrast=0.2),
    transforms.ToTensor(),
])
val_tf = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
])

# ----------------------------------------------------
# 4. Hierarchical Multi-Task Dataset
# ----------------------------------------------------
class HierarchicalPlantDataset(Dataset):
    def __init__(self, base_dir, common_labels, crop_to_idx, disease_to_idx, transform=None):
        self.files, self.crop_targets, self.health_targets, self.disease_targets = [], [], [], []
        self.transform = transform

        for cls in os.listdir(base_dir):
            cls_path = os.path.join(base_dir, cls)
            if not os.path.isdir(cls_path) or cls not in common_labels:
                continue

            crop_name, disease_name = split_label(cls)
            crop_id = crop_to_idx[crop_name]
            disease_id = disease_to_idx[disease_name]
            # Health label logic
            health_id = 1 if "healthy" in disease_name.lower() else 0

            for fname in os.listdir(cls_path):
                if fname.lower().endswith(IMG_EXTS):
                    self.files.append(os.path.join(cls_path, fname))
                    self.crop_targets.append(crop_id)
                    self.health_targets.append(health_id)
                    self.disease_targets.append(disease_id)

        print(f"Loaded {len(self.files)} images from {base_dir}")

    def __len__(self):
        return len(self.files)

    def __getitem__(self, idx):
        img = Image.open(self.files[idx]).convert("RGB")
        if self.transform:
            img = self.transform(img)
        return img, self.crop_targets[idx], self.health_targets[idx], self.disease_targets[idx]

# ----------------------------------------------------
# 5. MobileNetV3 Hierarchical MTL Model
# ----------------------------------------------------
class MobileNetV3_HierMTL(nn.Module):
    def __init__(self, num_crops, num_diseases):
        super().__init__()
        backbone = models.mobilenet_v3_large(weights="MobileNet_V3_Large_Weights.DEFAULT")
        feat_dim = backbone.classifier[0].in_features
        backbone.classifier = nn.Identity()
        self.backbone = backbone

        # Branch 1: Crop Classification
        self.crop_head = nn.Linear(feat_dim, num_crops)
        # Branch 2: Binary Health Classification (Healthy/Unhealthy)
        self.health_head = nn.Linear(feat_dim, 2)
        # Branch 3: Detailed Disease Classification
        self.disease_head = nn.Linear(feat_dim, num_diseases)

    def forward(self, x):
        feats = self.backbone(x)
        crop_logits = self.crop_head(feats)
        health_logits = self.health_head(feats)
        disease_logits = self.disease_head(feats)
        return crop_logits, health_logits, disease_logits, feats

# ----------------------------------------------------
# 6. CORAL Loss Function
# ----------------------------------------------------
def coral_loss(source_feats, target_feats):
    d = source_feats.size(1)
    src = source_feats - source_feats.mean(dim=0, keepdim=True)
    tgt = target_feats - target_feats.mean(dim=0, keepdim=True)
    cs = (src.t() @ src) / (source_feats.size(0) - 1 + 1e-6)
    ct = (tgt.t() @ tgt) / (target_feats.size(0) - 1 + 1e-6)
    return torch.mean((cs - ct) ** 2) / (4.0 * d * d)

# ----------------------------------------------------
# 7. Training Setup
# ----------------------------------------------------
train_dir = "/kaggle/input/plantvillage/PlantVillage/train"
val_dir = "/kaggle/input/plantvillage/PlantVillage/val"

train_ds = HierarchicalPlantDataset(train_dir, common_labels, crop_to_idx, disease_to_idx, transform=train_tf)
val_ds = HierarchicalPlantDataset(val_dir, common_labels, crop_to_idx, disease_to_idx, transform=val_tf)

train_loader = DataLoader(train_ds, batch_size=64, shuffle=True, num_workers=2)
val_loader = DataLoader(val_ds, batch_size=32, shuffle=False, num_workers=2)

model = MobileNetV3_HierMTL(num_crops, num_diseases).to(device)
optimizer = optim.Adam(model.parameters(), lr=1e-4)

# Losses
criterion_crop = nn.CrossEntropyLoss()
criterion_health = nn.CrossEntropyLoss()
criterion_disease = nn.CrossEntropyLoss()

# CORAL setup for Domain Adaptation
pd_unlabeled_dir = "/kaggle/working/PlantDoc-Dataset/test"
use_coral = False
tgt_iter = None
if os.path.isdir(pd_unlabeled_dir):
    # Simplified Unlabeled Dataset for CORAL
    class UnlabeledDataset(Dataset):
        def __init__(self, root, transform):
            self.files = [os.path.join(dp, f) for dp, dn, filenames in os.walk(root) for f in filenames if f.lower().endswith(IMG_EXTS)]
            self.transform = transform
        def __len__(self): return len(self.files)
        def __getitem__(self, idx):
            img = Image.open(self.files[idx]).convert("RGB")
            return self.transform(img) if self.transform else img

    tgt_ds = UnlabeledDataset(pd_unlabeled_dir, val_tf)
    if len(tgt_ds) > 0:
        tgt_loader = DataLoader(tgt_ds, batch_size=8, shuffle=True, num_workers=1)
        tgt_iter = cycle(tgt_loader)
        use_coral = True
        print("CORAL enabled with PlantDoc unlabeled data.")

# ----------------------------------------------------
# 8. Accuracy Helper
# ----------------------------------------------------
def get_accuracies(loader):
    model.eval()
    c_correct = h_correct = d_correct = total = 0
    with torch.no_grad():
        for imgs, crops, healths, diseases in loader:
            imgs, crops, healths, diseases = imgs.to(device), crops.to(device), healths.to(device), diseases.to(device)
            c_out, h_out, d_out, _ = model(imgs)

            c_correct += (c_out.argmax(1) == crops).sum().item()
            h_correct += (h_out.argmax(1) == healths).sum().item()
            d_correct += (d_out.argmax(1) == diseases).sum().item()
            total += crops.size(0)
    return c_correct/total, h_correct/total, d_correct/total

# ----------------------------------------------------
# 9. Training Loop
# ----------------------------------------------------
EPOCHS = 20
CORAL_LAMBDA = 0.1

for epoch in range(EPOCHS):
    model.train()
    running_loss = 0.0
    for imgs, c_target, h_target, d_target in train_loader:
        imgs, c_target, h_target, d_target = imgs.to(device), c_target.to(device), h_target.to(device), d_target.to(device)

        optimizer.zero_grad()
        c_out, h_out, d_out, feats_src = model(imgs)

        # Hierarchical Losses
        loss_crop = criterion_crop(c_out, c_target)
        loss_health = criterion_health(h_out, h_target)
        loss_disease = criterion_disease(d_out, d_target)

        loss = loss_crop + loss_health + loss_disease

        # Domain Adaptation (CORAL)
        if use_coral and tgt_iter:
            tgt_imgs = next(tgt_iter).to(device)
            _, _, _, feats_tgt = model(tgt_imgs)
            loss += CORAL_LAMBDA * coral_loss(feats_src, feats_tgt)

        loss.backward()
        optimizer.step()
        running_loss += loss.item()

    c_acc, h_acc, d_acc = get_accuracies(val_loader)
    print(f"Epoch {epoch+1}/{EPOCHS} | Loss: {running_loss/len(train_loader):.4f}")
    print(f"Val Acc -> Crop: {c_acc:.3f}, Health: {h_acc:.3f}, Disease: {d_acc:.3f}")

# 10. Save Model
torch.save({
    "state_dict": model.state_dict(),
    "crop_to_idx": crop_to_idx,
    "disease_to_idx": disease_to_idx,
}, "mobilenetv3_hierarchical_mtl_coral.pth")
print("Model saved successfully.")
```

```python
import os
import pandas as pd
from PIL import Image

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Device:", device)

# -----------------------------
# 0. Helpers
# -----------------------------
def split_label(full_label: str):
    return full_label.split("___")

IMG_EXTS = ('.jpg', '.jpeg', '.png', '.bmp', '.tif', '.tiff')

# -----------------------------
# 1. Load checkpoint + mappings (IMPORTANT)
# -----------------------------
ckpt_path = "mobilenetv3_hierarchical_mtl_coral.pth"
ckpt = torch.load(ckpt_path, map_location=device)

state_dict = ckpt["state_dict"] if "state_dict" in ckpt else ckpt
crop_to_idx = ckpt.get("crop_to_idx", None)
disease_to_idx = ckpt.get("disease_to_idx", None)

if crop_to_idx is None or disease_to_idx is None:
    raise ValueError("Checkpoint does not contain crop_to_idx and/or disease_to_idx. "
                     "Please save them during training or reconstruct identically.")

idx_to_crop = {i: c for c, i in crop_to_idx.items()}
idx_to_disease = {i: d for d, i in disease_to_idx.items()}

num_crops = len(crop_to_idx)
num_diseases = len(disease_to_idx)

print(f"Loaded mappings from ckpt: #Crops={num_crops}, #Diseases={num_diseases}")

# -----------------------------
# 2. Transforms (keep consistent with training/val)
# -----------------------------
val_tf = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
])

# -----------------------------
# 3. PlantDoc dataset (disease-only targets from mapping)
# -----------------------------
class PlantDocDiseaseDataset(Dataset):
    """
    Builds (image, disease_id) pairs using pd_to_pv_map:
      PlantDoc folder -> PlantVillage label (crop___disease)
    We then map disease to disease_id using disease_to_idx from checkpoint.
    """
    def __init__(self, pd_dir, transform, pd_to_pv_map, disease_to_idx, common_labels=None):
        self.files = []
        self.targets = []
        self.transform = transform

        for folder in os.listdir(pd_dir):
            if folder not in pd_to_pv_map:
                continue

            pv_label = pd_to_pv_map[folder]

            # optional: filter to known PV labels if you maintain common_labels
            if common_labels is not None and pv_label not in common_labels:
                continue

            _, disease_name = split_label(pv_label)
            if disease_name not in disease_to_idx:
                continue

            disease_id = disease_to_idx[disease_name]
            folder_path = os.path.join(pd_dir, folder)
            if not os.path.isdir(folder_path):
                continue

            for fname in os.listdir(folder_path):
                if fname.lower().endswith(IMG_EXTS):
                    self.files.append(os.path.join(folder_path, fname))
                    self.targets.append(disease_id)

        print(f"PlantDocDiseaseDataset: {len(self.files)} images from {pd_dir}")

    def __len__(self):
        return len(self.files)

    def __getitem__(self, idx):
        img = Image.open(self.files[idx]).convert("RGB")
        if self.transform:
            img = self.transform(img)
        label = self.targets[idx]
        return img, label

pd_test_dir = "/kaggle/working/PlantDoc-Dataset/test"
pd_dataset = PlantDocDiseaseDataset(
    pd_test_dir,
    val_tf,
    pd_to_pv_map=pd_to_pv_map,
    disease_to_idx=disease_to_idx,
    common_labels=common_labels  # if you have it; otherwise pass None
)
pd_loader = DataLoader(pd_dataset, batch_size=16, shuffle=False, num_workers=1)

# -----------------------------
# 4. Hierarchical MTL model definition (MUST match training)
# -----------------------------
class MobileNetV3_HierMTL(nn.Module):
    def __init__(self, num_crops, num_diseases):
        super().__init__()
        backbone = models.mobilenet_v3_large(weights="MobileNet_V3_Large_Weights.DEFAULT")
        feat_dim = backbone.classifier[0].in_features
        backbone.classifier = nn.Identity()
        self.backbone = backbone

        self.crop_head = nn.Linear(feat_dim, num_crops)
        self.health_head = nn.Linear(feat_dim, 2)          # healthy/unhealthy
        self.disease_head = nn.Linear(feat_dim, num_diseases)

    def forward(self, x):
        feats = self.backbone(x)
        crop_logits = self.crop_head(feats)
        health_logits = self.health_head(feats)
        disease_logits = self.disease_head(feats)
        return crop_logits, health_logits, disease_logits, feats

model = MobileNetV3_HierMTL(num_crops=num_crops, num_diseases=num_diseases).to(device)
model.load_state_dict(state_dict)
model.eval()
print("Loaded hierarchical MTL + CORAL model successfully.")

# -----------------------------
# 5. Evaluate disease accuracy + (optional) health accuracy
# -----------------------------
# disease_id -> true health label (0/1) inferred from disease name
disease_id_to_health = torch.tensor(
    [1 if "healthy" in idx_to_disease[i].lower() else 0 for i in range(num_diseases)],
    device=device
)

correct_dis = total = 0
correct_health = 0

all_true_dis_ids = []
all_pred_dis_ids = []
all_true_health = []
all_pred_health = []

with torch.no_grad():
    for images, true_dis in pd_loader:
        images = images.to(device)
        true_dis = true_dis.to(device)

        _, health_logits, disease_logits, _ = model(images)

        pred_dis = disease_logits.argmax(dim=1)
        pred_health = health_logits.argmax(dim=1)

        # disease accuracy
        correct_dis += (pred_dis == true_dis).sum().item()
        total += true_dis.size(0)

        # health accuracy (derived true health from disease label)
        true_health = disease_id_to_health[true_dis]
        correct_health += (pred_health == true_health).sum().item()

        all_true_dis_ids.extend(true_dis.cpu().tolist())
        all_pred_dis_ids.extend(pred_dis.cpu().tolist())
        all_true_health.extend(true_health.cpu().tolist())
        all_pred_health.extend(pred_health.cpu().tolist())

acc_dis = correct_dis / total if total > 0 else 0.0
acc_health = correct_health / total if total > 0 else 0.0

print(f"PlantDoc Test Disease Acc (HierMTL+CORAL): {acc_dis:.3f} ({correct_dis}/{total})")
print(f"PlantDoc Test Health  Acc (HierMTL+CORAL): {acc_health:.3f} ({correct_health}/{total})")

# -----------------------------
# 6. Save CSV with predictions
# -----------------------------
df = pd.DataFrame({
    "image_path": pd_dataset.files,
    "true_disease_id": all_true_dis_ids,
    "pred_disease_id": all_pred_dis_ids,
    "true_disease_name": [idx_to_disease[i] for i in all_true_dis_ids],
    "pred_disease_name": [idx_to_disease[i] for i in all_pred_dis_ids],
    "true_health": all_true_health,          # 0/1
    "pred_health": all_pred_health,          # 0/1
    "true_health_name": ["healthy" if x == 1 else "unhealthy" for x in all_true_health],
    "pred_health_name": ["healthy" if x == 1 else "unhealthy" for x in all_pred_health],
})

csv_path = "plantdoc_predictions_hier_mtl_coral.csv"
df.to_csv(csv_path, index=False)
print(f"Saved CSV to {csv_path}")
```

```python
import os
import torch
from torch import nn, optim
from torch.utils.data import DataLoader, Dataset
from torchvision import transforms, models
from PIL import Image
from itertools import cycle

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Device:", device)

# ----------------------------------------------------
# 0. Build crop & disease label spaces from common_labels
# ----------------------------------------------------
def split_label(full_label):
    return full_label.split("___")

crops = sorted({split_label(lbl)[0] for lbl in common_labels})
diseases = sorted({split_label(lbl)[1] for lbl in common_labels})

crop_to_idx = {c: i for i, c in enumerate(crops)}
disease_to_idx = {d: i for i, d in enumerate(diseases)}

num_crops = len(crops)
num_diseases = len(diseases)

print(f"#Crops: {num_crops}, #Diseases: {num_diseases}")

# Common image extensions
IMG_EXTS = ('.jpg', '.jpeg', '.png', '.bmp', '.tif', '.tiff')

# ----------------------------------------------------
# 1. Transforms (unchanged as per request)
# ----------------------------------------------------
train_tf = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(20),
    transforms.ColorJitter(brightness=0.2, contrast=0.2),
    transforms.ToTensor(),
])

val_tf = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
])

# ----------------------------------------------------
# 2. Multi-task PlantVillage dataset
# ----------------------------------------------------
class MultiTaskPlantVillageDataset(Dataset):
    def __init__(self, base_dir, common_labels, crop_to_idx, disease_to_idx, transform=None):
        self.files, self.crop_targets, self.disease_targets = [], [], []
        self.transform = transform

        for cls in os.listdir(base_dir):
            cls_path = os.path.join(base_dir, cls)
            if not os.path.isdir(cls_path) or cls not in common_labels:
                continue
            crop_name, disease_name = split_label(cls)
            crop_id = crop_to_idx[crop_name]
            disease_id = disease_to_idx[disease_name]
            for fname in os.listdir(cls_path):
                if fname.lower().endswith(IMG_EXTS):
                    self.files.append(os.path.join(cls_path, fname))
                    self.crop_targets.append(crop_id)
                    self.disease_targets.append(disease_id)

        print(f"Loaded {len(self.files)} images from {base_dir}")

    def __len__(self):
        return len(self.files)

    def __getitem__(self, idx):
        img = Image.open(self.files[idx]).convert("RGB")
        if self.transform:
            img = self.transform(img)
        return img, self.crop_targets[idx], self.disease_targets[idx]

train_dir = "/kaggle/input/plantvillage/PlantVillage/train"
val_dir   = "/kaggle/input/plantvillage/PlantVillage/val"

train_ds = MultiTaskPlantVillageDataset(train_dir, common_labels, crop_to_idx, disease_to_idx, transform=train_tf)
val_ds   = MultiTaskPlantVillageDataset(val_dir,   common_labels, crop_to_idx, disease_to_idx, transform=val_tf)

train_loader = DataLoader(train_ds, batch_size=64, shuffle=True, num_workers=1)
val_loader   = DataLoader(val_ds, batch_size=32, shuffle=False, num_workers=1)

# ----------------------------------------------------
# 3. IMPROVED: Multi-scale MobileNetV3 MTL model
# ----------------------------------------------------
class MobileNetV3_MTL(nn.Module):
    def __init__(self, num_crops, num_diseases):
        super().__init__()
        backbone = models.mobilenet_v3_large(weights="MobileNet_V3_Large_Weights.DEFAULT")

        # Split backbone into stages for multi-scale features
        self.stage1 = nn.Sequential(*list(backbone.features[:7]))   # Early features
        self.stage2 = nn.Sequential(*list(backbone.features[7:13])) # Mid features
        self.stage3 = nn.Sequential(*list(backbone.features[13:]))  # High features
        self.avgpool = backbone.avgpool

        feat_dim = 960
        self.crop_head = nn.Linear(feat_dim, num_crops)
        self.disease_head = nn.Linear(feat_dim, num_diseases)

    def forward(self, x, return_multi_scale=False):
        f1 = self.stage1(x)
        f2 = self.stage2(f1)
        f3 = self.stage3(f2)
        feats = self.avgpool(f3).flatten(1)

        crop_logits = self.crop_head(feats)
        disease_logits = self.disease_head(feats)

        if return_multi_scale:
            # Return flattened intermediate features for multi-scale CORAL
            f2_flat = torch.nn.functional.adaptive_avg_pool2d(f2, 1).flatten(1)
            return crop_logits, disease_logits, feats, f2_flat
        return crop_logits, disease_logits, feats

# ----------------------------------------------------
# 4. CORAL loss (unchanged)
# ----------------------------------------------------
def coral_loss(source_feats, target_feats):
    d = source_feats.size(1)
    src = source_feats - source_feats.mean(dim=0, keepdim=True)
    tgt = target_feats - target_feats.mean(dim=0, keepdim=True)
    cs = (src.t() @ src) / (source_feats.size(0) - 1)
    ct = (tgt.t() @ tgt) / (target_feats.size(0) - 1)
    return torch.mean((cs - ct) ** 2) / (4.0 * d * d)

# ----------------------------------------------------
# 5. NEW: Entropy minimization loss
# ----------------------------------------------------
def entropy_loss(logits):
    """Minimize prediction entropy on unlabeled target data"""
    p = torch.softmax(logits, dim=1)
    log_p = torch.log_softmax(logits, dim=1)
    entropy = -torch.sum(p * log_p, dim=1)
    return entropy.mean()

# ----------------------------------------------------
# 6. Model, loss, optimizer
# ----------------------------------------------------
model_mtl = MobileNetV3_MTL(num_crops, num_diseases).to(device)
criterion_crop = nn.CrossEntropyLoss()
criterion_disease = nn.CrossEntropyLoss()

# IMPROVED: Better optimizer and hyperparameters
optimizer = optim.AdamW(model_mtl.parameters(), lr=3e-4, weight_decay=1e-4)
scheduler = optim.lr_scheduler.CosineAnnealingWarmRestarts(optimizer, T_0=5, T_mult=2)

# IMPROVED: Better hyperparameters
CORAL_LAMBDA = 0.3      # Increased from 0.1
ENTROPY_LAMBDA = 0.1    # New
PSEUDO_LAMBDA = 0.5     # For pseudo-labeling
PSEUDO_THRESHOLD = 0.9  # Confidence threshold
TGT_BATCH_SIZE = 64     # Match source batch size

# ----------------------------------------------------
# 7. Unlabeled PlantDoc dataset for CORAL
# ----------------------------------------------------
pd_unlabeled_dir = "/kaggle/working/PlantDoc-Dataset/test"

use_coral = False
tgt_iter = None

if os.path.isdir(pd_unlabeled_dir):
    class UnlabeledPlantDocDataset(Dataset):
        def __init__(self, root, transform):
            self.files = []
            self.transform = transform

            for cls in os.listdir(root):
                cls_path = os.path.join(root, cls)
                if not os.path.isdir(cls_path):
                    continue
                for fname in os.listdir(cls_path):
                    if fname.lower().endswith(IMG_EXTS):
                        self.files.append(os.path.join(cls_path, fname))

            print(f"Loaded {len(self.files)} unlabeled images from PlantDoc test for CORAL.")

        def __len__(self):
            return len(self.files)

        def __getitem__(self, idx):
            img = Image.open(self.files[idx]).convert("RGB")
            if self.transform:
                img = self.transform(img)
            return img

    tgt_ds = UnlabeledPlantDocDataset(pd_unlabeled_dir, val_tf)
    if len(tgt_ds) > 0:
        tgt_loader = DataLoader(tgt_ds, batch_size=TGT_BATCH_SIZE, shuffle=True, num_workers=1)
        tgt_iter = cycle(tgt_loader)
        use_coral = True
        print("CORAL enabled.")
    else:
        print("PlantDoc test folder has no images – CORAL disabled.")
else:
    print("pd_unlabeled_dir does not exist – CORAL disabled.")

# ----------------------------------------------------
# 8. Accuracy helper
# ----------------------------------------------------
def mtl_accuracy(loader):
    model_mtl.eval()
    correct_crop = total_crop = 0
    correct_dis = total_dis = 0
    with torch.no_grad():
        for images, crop_labels, disease_labels in loader:
            images = images.to(device)
            crop_labels = crop_labels.to(device)
            disease_labels = disease_labels.to(device)

            crop_logits, disease_logits, _ = model_mtl(images)
            correct_crop += (crop_logits.argmax(1) == crop_labels).sum().item()
            total_crop += crop_labels.size(0)

            correct_dis += (disease_logits.argmax(1) == disease_labels).sum().item()
            total_dis += disease_labels.size(0)

    return correct_crop / total_crop, correct_dis / total_dis

# ----------------------------------------------------
# 9. IMPROVED: Training loop with multi-scale CORAL,
#    entropy minimization, and pseudo-labeling
# ----------------------------------------------------
EPOCHS = 20  # Increased from 1

for epoch in range(EPOCHS):
    model_mtl.train()
    running_loss = 0.0
    running_coral = 0.0
    running_entropy = 0.0
    running_pseudo = 0.0

    for images, crop_labels, disease_labels in train_loader:
        images = images.to(device)
        crop_labels = crop_labels.to(device)
        disease_labels = disease_labels.to(device)

        optimizer.zero_grad()

        # Source domain forward pass with multi-scale features
        crop_logits, disease_logits, feats_src, f2_src = model_mtl(images, return_multi_scale=True)

        # Supervised losses
        loss_crop = criterion_crop(crop_logits, crop_labels)
        loss_disease = criterion_disease(disease_logits, disease_labels)
        loss = loss_crop + loss_disease

        # Domain adaptation on target domain
        if use_coral and tgt_iter is not None:
            tgt_imgs = next(tgt_iter).to(device)

            # FIXED: Remove torch.no_grad() to allow gradient flow
            crop_logits_tgt, disease_logits_tgt, feats_tgt, f2_tgt = model_mtl(
                tgt_imgs, return_multi_scale=True
            )

            # Match batch sizes
            min_size = min(feats_src.size(0), feats_tgt.size(0))

            # Multi-scale CORAL loss
            loss_coral_high = coral_loss(feats_src[:min_size], feats_tgt[:min_size])
            loss_coral_mid = coral_loss(f2_src[:min_size], f2_tgt[:min_size])
            loss_coral_total = loss_coral_high + 0.5 * loss_coral_mid

            # Entropy minimization on target predictions
            loss_ent = entropy_loss(crop_logits_tgt) + entropy_loss(disease_logits_tgt)

            loss += CORAL_LAMBDA * loss_coral_total + ENTROPY_LAMBDA * loss_ent
            running_coral += loss_coral_total.item()
            running_entropy += loss_ent.item()

            # Pseudo-labeling (start after warmup)
            if epoch >= 3:
                with torch.no_grad():
                    crop_probs = torch.softmax(crop_logits_tgt, dim=1)
                    disease_probs = torch.softmax(disease_logits_tgt, dim=1)

                    crop_conf, crop_pseudo = crop_probs.max(1)
                    disease_conf, disease_pseudo = disease_probs.max(1)

                    crop_mask = crop_conf > PSEUDO_THRESHOLD
                    disease_mask = disease_conf > PSEUDO_THRESHOLD

                # Apply pseudo-labels for high-confidence predictions
                loss_pseudo = 0.0
                if crop_mask.sum() > 0:
                    loss_pseudo += criterion_crop(crop_logits_tgt[crop_mask], crop_pseudo[crop_mask])

                if disease_mask.sum() > 0:
                    loss_pseudo += criterion_disease(disease_logits_tgt[disease_mask], disease_pseudo[disease_mask])

                if crop_mask.sum() > 0 or disease_mask.sum() > 0:
                    loss += PSEUDO_LAMBDA * loss_pseudo
                    running_pseudo += loss_pseudo.item()

        loss.backward()

        # Gradient clipping for stability
        torch.nn.utils.clip_grad_norm_(model_mtl.parameters(), 1.0)

        optimizer.step()
        running_loss += loss.item()

    scheduler.step()

    # Evaluate
    train_crop_acc, train_dis_acc = mtl_accuracy(train_loader)
    val_crop_acc, val_dis_acc = mtl_accuracy(val_loader)

    print(f"Epoch {epoch+1}/{EPOCHS} | "
          f"Loss: {running_loss/len(train_loader):.4f} | "
          f"CORAL: {running_coral/len(train_loader):.4f} | "
          f"Entropy: {running_entropy/len(train_loader):.4f} | "
          f"Pseudo: {running_pseudo/len(train_loader):.4f} | "
          f"LR: {scheduler.get_last_lr()[0]:.2e} | "
          f"Train Crop: {train_crop_acc:.3f} | Train Dis: {train_dis_acc:.3f} | "
          f"Val Crop: {val_crop_acc:.3f} | Val Dis: {val_dis_acc:.3f}")

# ----------------------------------------------------
# 10. Save model
# ----------------------------------------------------
torch.save({
    "state_dict": model_mtl.state_dict(),
    "crop_to_idx": crop_to_idx,
    "disease_to_idx": disease_to_idx,
}, "mobilenetv3_mtl_coral_improved.pth")

print("Improved MTL + CORAL model saved as mobilenetv3_mtl_coral_improved.pth")
```

	•	Keeps your corrected MTL model (EfficientNet-B0 backbone, conditional disease head, label smoothing, per-sample class weights).
    
	•	Adds an unsupervised PlantDoc (PD) loader for CORAL (no PD labels required).
    
	•	Computes CORAL on backbone features (feat = pool(backbone(x)).flatten(1)).
    
	•	Mixes supervised PV task loss + λ * CORAL_loss.

	•	Handles unequal source/target loader lengths by cycling the smaller loader.
    
	•	Keeps gradient clipping and evaluation hooks.

class LabelSmoothingCE(nn.Module):
    def __init__(self, smoothing=0.1):
        super().__init__()
        self.smoothing = smoothing
    def forward(self, logits, target, weight=None):
        num_classes = logits.size(1)
        log_probs = F.log_softmax(logits, dim=1)
        with torch.no_grad():
            true_dist = torch.full_like(log_probs, self.smoothing / (num_classes - 1))
            true_dist.scatter_(1, target.unsqueeze(1), 1.0 - self.smoothing)
        per_sample_loss = - (true_dist * log_probs).sum(dim=1)
        if weight is not None:
            sample_weight = weight[target]
            per_sample_loss = per_sample_loss * sample_weight
        return per_sample_loss.mean()

def coral_loss(source_feats, target_feats):
    d = source_feats.size(1)
    src = source_feats - source_feats.mean(dim=0, keepdim=True)
    tgt = target_feats - target_feats.mean(dim=0, keepdim=True)
    cs = (src.t() @ src) / (source_feats.size(0) - 1)
    ct = (tgt.t() @ tgt) / (target_feats.size(0) - 1)
    loss = torch.mean((cs - ct)**2) / (4.0 * d * d)
    return loss

# ----------------------------------------------------
# 4. Hierarchical MTL Model: Disease → Crop
# ----------------------------------------------------
class MobileNetV3_HierMTL(nn.Module):
    """
    Hierarchical MTL:
      Backbone -> Disease head -> Crop head

    Disease is the PRIMARY task.
    Crop head takes disease logits as input (disease → crop).
    We detach the disease logits so crop does NOT affect backbone/disease.
    """
    def __init__(self, num_crops, num_diseases):
        super().__init__()
        backbone = models.mobilenet_v3_large(
            weights=models.MobileNet_V3_Large_Weights.DEFAULT
        )
        # feature dim = input to classifier[0] (960 for MobileNetV3-Large)
        feat_dim = backbone.classifier[0].in_features
        backbone.classifier = nn.Identity()
        self.backbone = backbone

        # primary head: disease
        self.disease_head = nn.Linear(feat_dim, num_diseases)
        # secondary head: crop, on top of disease logits
        self.crop_head = nn.Linear(num_diseases, num_crops)

    def forward(self, x):
        feats = self.backbone(x)                 # [B, 960]
        disease_logits = self.disease_head(feats)  # [B, num_diseases]

        # Hierarchical: crop depends on disease
        # detach so crop doesn't backprop into disease/backbone
        crop_logits = self.crop_head(disease_logits.detach())  # [B, num_crops]

        return crop_logits, disease_logits

model_hier = MobileNetV3_HierMTL(num_crops=num_crops, num_diseases=num_diseases).to(device)

# ----------------------------------------------------
# 5. Loss & optimizer (Disease primary, Crop auxiliary)
# ----------------------------------------------------
criterion_crop = nn.CrossEntropyLoss()
criterion_disease = nn.CrossEntropyLoss()

# Strongly prioritize disease for domain shift
crop_loss_weight = 0.01
disease_loss_weight = 1.0

optimizer = optim.Adam(model_hier.parameters(), lr=1e-4)

# ----------------------------------------------------
# 6. Accuracy helpers
# ----------------------------------------------------
def hier_accuracy(loader):
    model_hier.eval()
    correct_crop = total_crop = 0
    correct_disease = total_disease = 0

    with torch.no_grad():
        for images, crop_labels, disease_labels in loader:
            images = images.to(device)
            crop_labels = crop_labels.to(device)
            disease_labels = disease_labels.to(device)

            crop_logits, disease_logits = model_hier(images)
            crop_preds = crop_logits.argmax(1)
            disease_preds = disease_logits.argmax(1)

            correct_crop += (crop_preds == crop_labels).sum().item()
            total_crop   += crop_labels.size(0)

            correct_disease += (disease_preds == disease_labels).sum().item()
            total_disease   += disease_labels.size(0)

    return (correct_crop / total_crop,
            correct_disease / total_disease)

# ----------------------------------------------------
# 7. Training loop (Hierarchical MTL)
# ----------------------------------------------------
EPOCHS = 20 # adjust as needed

for epoch in range(EPOCHS):
    model_hier.train()
    running_loss = 0.0

    for images, crop_labels, disease_labels in train_loader:
        images = images.to(device)
        crop_labels = crop_labels.to(device)
        disease_labels = disease_labels.to(device)

        optimizer.zero_grad()

        crop_logits, disease_logits = model_hier(images)

        loss_crop = criterion_crop(crop_logits, crop_labels)
        loss_disease = criterion_disease(disease_logits, disease_labels)

        # disease is primary
        loss = crop_loss_weight * loss_crop + disease_loss_weight * loss_disease
        loss.backward()
        optimizer.step()

        running_loss += loss.item()

    train_crop_acc, train_dis_acc = hier_accuracy(train_loader)
    val_crop_acc, val_dis_acc     = hier_accuracy(val_loader)

    print(f"Epoch {epoch+1}/{EPOCHS} | "
          f"Loss: {running_loss/len(train_loader):.4f} | "
          f"Train Crop Acc: {train_crop_acc:.3f} | Train Dis Acc: {train_dis_acc:.3f} | "
          f"Val Crop Acc: {val_crop_acc:.3f} | Val Dis Acc: {val_dis_acc:.3f}")
